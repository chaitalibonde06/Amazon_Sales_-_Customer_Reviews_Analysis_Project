```python
# Step 4 - Data Cleaning : 
# Correcting 'rating' column : 
# finding incorrect values : 
df[df['rating'] == '|']
```

```python
# 1. Fix the rating column :
# Replace '|' with a missing value (NaN)
df['rating'] = df['rating'].replace('|', np.nan)

# changing datatype of 'rating' column 
df['rating'] = pd.to_numeric(df['rating'],errors = 'coerce')

# Filling missing values with median 
df['rating'] = df['rating'].fillna(df['rating'].median())

# checking missing values and datatype : 
print(df['rating'].isnull().sum())
print(df['rating'].dtypes)
```

```python
# 2. Cleaning all price columns : 
# Remove the ₹ symbol and commas
df['discounted_price'] = df['discounted_price'].str.replace('₹','').str.replace(',','')
# Chaging datatype of column 
df['discounted_price'] = pd.to_numeric(df['discounted_price'],errors = 'coerce')

df['actual_price'] = df['actual_price'].str.replace('₹','').str.replace(',','')
df['actual_price'] = pd.to_numeric(df['actual_price'], errors = 'coerce')
```

```python
# cleaning 'discount_percentage' column : 
# remove '%' symbol and changing the datatype : 
df['discount_percentage'] = df['discount_percentage'].str.replace('%','')
df['discount_percentage'] = pd.to_numeric(df['discount_percentage'], errors = 'coerce')
```

```python
print(df[['discounted_price', 'actual_price', 'discount_percentage']].isnull().sum())
print(df[['discounted_price', 'actual_price', 'discount_percentage']].dtypes)
df[['discounted_price', 'actual_price', 'discount_percentage']].head()
```

```python
# 3. Find missing values from 'rating_count' column :
df[df['rating_count'].isnull()]
```

```python
# clean the 'rating_count' column : 
# remove commas and change the datatype 
df['rating_count'] = df['rating_count'].str.replace(',','')
df['rating_count'] = pd.to_numeric(df['rating_count'], errors = 'coerce')

# fill missing values with median 
df['rating_count'] = df['rating_count'].fillna(df['rating_count'].median())
```

```python
print(df['rating_count'].isnull().sum())
print(df['rating_count'].dtypes)
```

```python
# 4. check repeated rows : 
# show all rows whose product_id appears more than once 
df[df['product_id'].duplicated(keep=False)].sort_values('product_id').head(10)
```

```python
# remove repeated rows : 
# (we can keep first row of each product_id) 
df = df.drop_duplicates(subset='product_id', keep='first')

# reset the index after removing rows 
df = df.reset_index(drop = True)
```

```python
print(df.shape)
print(df['product_id'].duplicated().sum())
```

```python
# Task 4 : Analyze product category distribution -
#  5 - Extracting main_category from 'category' column :
# Create main_category and sub_category 
# split the category text at '|' and take the first part as main_category :
df['main_category'] = df['category'].str.split('|').str[0]

# take the second part as sub_category 
df['sub_category'] = df['category'].str.split('|').str[1]
```

```python
print(df[['category', 'main_category', 'sub_category']].head())

# count products in each main category 
print(df['main_category'].value_counts())
```

```python
# 6. Create 'discount_amount' column : 
df['discount_amount'] = df['actual_price'] - df['discounted_price']
```

```python
# 7. Create 'price_range' column :
# divide products into three groups based on discounted price 
df['price_range'] = pd.cut(
    df['discounted_price'],
    bins = [0, 500, 2000, float('inf')],
    labels = ['Budget', 'Mid-Range', 'Premium']
)
```

```python
# show ghe new columns : 
print(df[['actual_price', 'discounted_price', 'discount_amount', 'price_range']].head())

# count products in each price range
print(df['price_range'].value_counts())
```

```python
# Create total_sales column : 
# Estimated sales = discounted price x number of ratings
df['total_sales'] = df['discounted_price'] * df['rating_count']
```

```python
# Check the result
# Show the new column
print(df[['product_name', 'discounted_price', 'rating_count', 'total_sales']].head())

# Total estimated sales of all products
print("Total sales:", df['total_sales'].sum())
```

## Task 10: Business Insights and Recommendations

### Key Findings
1. **Category focus:** Electronics (490), Home&Kitchen (448) and Computers&Accessories (375) make up about 97% of all products. The catalog is heavily concentrated in these three categories.
2. **Discounts:** The average discount is about 47%, and most products are discounted between 40% and 60%. Budget products receive the highest average discount (about 55%), while Premium products receive the lowest (about 37%).
3. **Ratings:** Customer ratings are generally positive, with an average of about 4.1 and around 75% of products rated 4.0 or above. Ratings are very similar across categories and price ranges.
4. **Best sellers:** The most popular products by rating count are low-priced Electronics such as HDMI cables, earphones and entry-level smartphones, and they also have strong ratings (4.1 to 4.4).
5. **Correlation:** Price, actual price and discount amount are strongly related to each other. Rating and rating count have only a weak relationship with price and discount, so a higher discount does not lead to higher ratings or more customer engagement.

### Recommendations
1. **Prioritize the top three categories:** Focus marketing, stock and offers on Electronics, Home&Kitchen and Computers&Accessories, because they hold almost all products and customer activity.
2. **Review the discount strategy:** Since higher discounts do not improve ratings or popularity, avoid unnecessary deep discounts, especially on products that already sell well.
3. **Promote popular everyday products:** Cables, earphones and entry-level phones attract the most customer engagement, so they should be featured and always kept in stock.
4. **Improve weaker categories:** Home&Kitchen has the lowest average rating among the three large categories (4.04), so product quality and product information in this category should be reviewed.
5. **Clean up product variants:** Many popular products are listed several times as separate variants. Grouping them would make sales and customer feedback easier to track.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")


def save_chart(filename):
    # Save the current chart as a PNG file and close it
    plt.tight_layout()
    plt.savefig(f"report_output/charts/{filename}", dpi=150, bbox_inches="tight")
    plt.close()
    print("Saved:", filename)


# 1. Product Category Count Plot (palette: viridis)
counts = rdf['main_category'].value_counts()
plt.figure(figsize=(10, 5))
sns.barplot(x=counts.values, y=counts.index, hue=counts.index, palette='viridis', legend=False)
plt.title('Number of Products in Each Main Category')
plt.xlabel('Number of Products')
plt.ylabel('Main Category')
save_chart("01_category_count.png")

# 2. Product Rating Distribution (color: orange)
plt.figure(figsize=(10, 5))
sns.histplot(rdf['rating'], bins=15, color='orange')
plt.title('Distribution of Product Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Products')
save_chart("02_rating_distribution.png")

# 3. Discount Histogram (color: teal)
plt.figure(figsize=(10, 5))
sns.histplot(rdf['discount_percentage'], bins=20, color='teal')
plt.title('Distribution of Discount Percentage')
plt.xlabel('Discount Percentage (%)')
plt.ylabel('Number of Products')
save_chart("03_discount_histogram.png")

# 4. Price Distribution Histogram, products up to Rs 10,000 (color: teal)
plt.figure(figsize=(10, 5))
sns.histplot(rdf[rdf['discounted_price'] <= 10000]['discounted_price'], bins=40, color='teal')
plt.title('Price Distribution of Products up to Rs 10,000')
plt.xlabel('Discounted Price (Rs)')
plt.ylabel('Number of Products')
save_chart("04_price_histogram.png")

# 5. Top Products Bar Chart (palette: crest)
top = rdf.sort_values('rating_count', ascending=False).head(10).copy()
top['short_name'] = top['product_name'].str[:25] + ' (' + top['product_id'].str[-4:] + ')'
plt.figure(figsize=(10, 6))
sns.barplot(x=top['rating_count'], y=top['short_name'], hue=top['short_name'], palette='crest', legend=False)
plt.title('Top 10 Best-Selling Products (by Rating Count)')
plt.xlabel('Rating Count')
plt.ylabel('Product Name')
save_chart("05_top_products.png")

# 6. Ratings vs Price Scatter Plot (color: purple)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=rdf, x='discounted_price', y='rating', alpha=0.5, color='purple')
plt.xscale('log')
plt.title('Rating vs Discounted Price')
plt.xlabel('Discounted Price (Rs, log scale)')
plt.ylabel('Rating')
save_chart("06_rating_vs_price.png")

# 7. Discount vs Sales Scatter Plot (color: green)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=rdf, x='discount_percentage', y='total_sales', alpha=0.5, color='green')
plt.yscale('log')
plt.title('Discount Percentage vs Estimated Sales')
plt.xlabel('Discount Percentage (%)')
plt.ylabel('Estimated Sales (Rs, log scale)')
save_chart("07_discount_vs_sales.png")

# 8. Box Plot of Product Prices (color: lightgreen)
plt.figure(figsize=(6, 6))
sns.boxplot(y=rdf['discounted_price'], color='lightgreen')
plt.yscale('log')
plt.title('Box Plot of Product Prices')
plt.ylabel('Discounted Price (Rs, log scale)')
save_chart("08_price_boxplot.png")

# 9. Correlation Heatmap (cmap: coolwarm)
corr_cols = ['discounted_price', 'actual_price', 'discount_amount',
             'discount_percentage', 'rating', 'rating_count']
plt.figure(figsize=(9, 6))
sns.heatmap(rdf[corr_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Correlation Heatmap of Numeric Columns')
save_chart("09_correlation_heatmap.png")

# 10. Pair Plot (default colors, log values for price and rating count)
pair_df = rdf[['discounted_price', 'discount_percentage', 'rating', 'rating_count']].copy()
pair_df['discounted_price'] = np.log10(pair_df['discounted_price'])
pair_df['rating_count'] = np.log10(pair_df['rating_count'])
pair_df.columns = ['log_price', 'discount_percentage', 'rating', 'log_rating_count']
g = sns.pairplot(pair_df, plot_kws={'alpha': 0.4, 's': 20}, diag_kind='hist')
g.fig.suptitle('Pair Plot of Price, Discount, Rating and Rating Count', y=1.02)
g.savefig("report_output/charts/10_pair_plot.png", dpi=120, bbox_inches="tight")
plt.close()
print("Saved: 10_pair_plot.png")
```

```python
# ---------- Calculate the numbers ----------
n_products = len(rdf)
total_rev = rdf['total_sales'].sum()
avg_rating = rdf['rating'].mean()
avg_disc = rdf['discount_percentage'].mean()
total_reviews = rdf['rating_count'].sum()

# Category insights
cat_counts = rdf['main_category'].value_counts()
top3_share = cat_counts.head(3).sum() / n_products * 100
cat_rev = rdf.groupby('main_category')['total_sales'].sum().sort_values(ascending=False)
top_cat = cat_rev.index[0]
top_cat_share = cat_rev.iloc[0] / total_rev * 100

# Discount insights
disc_by_price = rdf.groupby('price_range', observed=True)['discount_percentage'].mean()
high_disc_range = disc_by_price.idxmax()
low_disc_range = disc_by_price.idxmin()
disc_sales_corr = rdf['discount_percentage'].corr(rdf['total_sales'])

# Rating insights
share_high_rated = (rdf['rating'] >= 4).mean() * 100
low_rated = (rdf['rating'] < 3).sum()
big_cats = cat_counts[cat_counts >= 30].index
rating_by_cat = rdf[rdf['main_category'].isin(big_cats)].groupby('main_category')['rating'].mean()
best_cat = rating_by_cat.idxmax()
worst_cat = rating_by_cat.idxmin()
price_rating_corr = rdf['discounted_price'].corr(rdf['rating'])

# Best-selling product and reviews
best = rdf.sort_values('total_sales', ascending=False).iloc[0]
best_name = best['product_name'].split('(')[0].strip()
cat_reviews = rdf.groupby('main_category')['rating_count'].sum().sort_values(ascending=False)
top_review_cat = cat_reviews.index[0]
top_review_share = cat_reviews.iloc[0] / total_reviews * 100

# Price insights
median_price = rdf['discounted_price'].median()
share_under_2000 = (rdf['discounted_price'] <= 2000).mean() * 100


# ---------- Helper: describe a correlation in words ----------
def link_text(c):
    if abs(c) < 0.3:
        return "there is no strong relationship between the two"
    if c > 0:
        return "the two increase together"
    return "when one increases the other tends to decrease"


# ---------- Write the insights as sentences ----------
insights = [
    f"The dataset contains {n_products:,} unique products with an estimated revenue of Rs {total_rev/1e9:.2f} billion, an average rating of {avg_rating:.2f} and an average discount of {avg_disc:.1f}%.",
    f"The top three categories ({', '.join(cat_counts.head(3).index)}) hold {top3_share:.1f}% of all products, so the catalog is concentrated in a few categories.",
    f"{top_cat} generates {top_cat_share:.1f}% of the total estimated revenue, and {top_review_cat} receives {top_review_share:.1f}% of all customer reviews.",
    f"{best_name} is the highest-selling product with an estimated revenue of Rs {best['total_sales']/1e9:.2f} billion.",
    f"Customer ratings are positive: {share_high_rated:.1f}% of products are rated 4.0 or above, and only {low_rated} products are rated below 3.",
    f"Among large categories, {best_cat} has the highest average rating ({rating_by_cat.max():.2f}) and {worst_cat} has the lowest ({rating_by_cat.min():.2f}).",
    f"{high_disc_range} products get the highest average discount ({disc_by_price.max():.1f}%), while {low_disc_range} products get the lowest ({disc_by_price.min():.1f}%).",
    f"The correlation between discount percentage and estimated sales is {disc_sales_corr:.2f}, so {link_text(disc_sales_corr)}.",
    f"The correlation between price and rating is {price_rating_corr:.2f}, so {link_text(price_rating_corr)}.",
    f"The median product price is Rs {median_price:,.0f}, and {share_under_2000:.1f}% of products cost Rs 2,000 or less.",
]

# Print the insights and save them in a text file
for i, line in enumerate(insights, start=1):
    print(f"{i}. {line}")

with open("report_output/business_insights.txt", "w", encoding="utf-8") as f:
    f.write("BUSINESS INSIGHTS\n\n")
    for i, line in enumerate(insights, start=1):
        f.write(f"{i}. {line}\n")
```

```python
# ---------- Write the recommendations as sentences ----------
recommendations = [
    f"Prioritize the top categories: focus stock, marketing and offers on {', '.join(cat_counts.head(3).index)}, which hold {top3_share:.0f}% of all products, and {top_cat} alone brings {top_cat_share:.0f}% of the revenue.",
    f"Review the discount strategy: the correlation between discount and sales is {disc_sales_corr:.2f}, so avoid unnecessary deep discounts, especially on {high_disc_range} products (average discount {disc_by_price.max():.0f}%) and on products that already sell well.",
    f"Promote best sellers: keep top products such as {best_name} always in stock and feature them prominently.",
    f"Improve weaker areas: review the {low_rated} products rated below 3, and check product quality and product information in {worst_cat}, which has the lowest average rating ({rating_by_cat.min():.2f}) among the large categories.",
    f"Expand the smaller categories: all categories outside the top three hold only {100 - top3_share:.1f}% of products, so adding more products there could widen the catalog.",
    "Clean up product listings: many popular products are listed several times as separate variants, so grouping them would make sales and customer feedback easier to track.",
]

# Print the recommendations and save them in a text file
for i, line in enumerate(recommendations, start=1):
    print(f"{i}. {line}")

with open("report_output/recommendations.txt", "w", encoding="utf-8") as f:
    f.write("FINAL RECOMMENDATIONS\n\n")
    for i, line in enumerate(recommendations, start=1):
        f.write(f"{i}. {line}\n")
```

```python
# ---------- Step 7: Build the PDF report ----------
from datetime import date
from reportlab.lib.pagesizes import A4
from reportlab.lib import colors
from reportlab.lib.units import cm
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.platypus import (SimpleDocTemplate, Paragraph, Spacer, Image,
                                Table, TableStyle, PageBreak, KeepTogether)
from PIL import Image as PILImage
from xml.sax.saxutils import escape   # makes '&' safe inside PDF text

NAVY_C, BLUE_C, ORANGE_C = colors.HexColor("#232F3E"), colors.HexColor("#3B82A0"), colors.HexColor("#FF9900")

styles = getSampleStyleSheet()
title_s = ParagraphStyle("t", parent=styles["Title"], textColor=NAVY_C, fontSize=26, leading=32)
sub_s   = ParagraphStyle("s", parent=styles["Normal"], textColor=colors.grey, fontSize=12, alignment=1)
h1      = ParagraphStyle("h1", parent=styles["Heading1"], textColor=NAVY_C, fontSize=18, spaceAfter=8)
body    = ParagraphStyle("b", parent=styles["BodyText"], fontSize=10.5, leading=15, spaceAfter=6)
cap_s   = ParagraphStyle("c", parent=styles["Italic"], fontSize=9.5, textColor=colors.HexColor("#555555"), spaceAfter=10)


def header_footer(canvas, doc):
    canvas.saveState()
    canvas.setFillColor(NAVY_C)
    canvas.rect(0, A4[1] - 1.0 * cm, A4[0], 1.0 * cm, fill=1, stroke=0)
    canvas.setFillColor(ORANGE_C)
    canvas.rect(0, A4[1] - 1.1 * cm, A4[0], 0.1 * cm, fill=1, stroke=0)
    canvas.setFillColor(colors.grey)
    canvas.setFont("Helvetica", 8)
    canvas.drawString(2 * cm, 1 * cm, "Amazon Product Sales & Customer Reviews Analysis")
    canvas.drawRightString(A4[0] - 2 * cm, 1 * cm, f"Page {doc.page}")
    canvas.restoreState()


def chart_image(path, max_w=16.5 * cm, max_h=10.5 * cm):
    # Keep the original proportions of each chart
    w, h = PILImage.open(path).size
    scale = min(max_w / w, max_h / h)
    return Image(path, width=w * scale, height=h * scale)


def make_table(data, col_widths=None, font=9):
    t = Table(data, colWidths=col_widths, repeatRows=1)
    t.setStyle(TableStyle([
        ("BACKGROUND", (0, 0), (-1, 0), NAVY_C),
        ("TEXTCOLOR", (0, 0), (-1, 0), colors.white),
        ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
        ("FONTSIZE", (0, 0), (-1, -1), font),
        ("ALIGN", (1, 0), (-1, -1), "RIGHT"),
        ("ROWBACKGROUNDS", (0, 1), (-1, -1), [colors.white, colors.HexColor("#F5F7FA")]),
        ("GRID", (0, 0), (-1, -1), 0.3, colors.HexColor("#D9DEE5")),
        ("TOPPADDING", (0, 0), (-1, -1), 5), ("BOTTOMPADDING", (0, 0), (-1, -1), 5),
    ]))
    return t


story = []

# ----- Cover page -----
story += [Spacer(1, 5 * cm),
          Paragraph("Amazon Product Sales &amp;<br/>Customer Reviews Analysis", title_s),
          Spacer(1, 0.5 * cm),
          Paragraph("Automatically generated report - Project 9", sub_s),
          Spacer(1, 0.3 * cm),
          Paragraph(f"Generated on {date.today().strftime('%d %B %Y')}", sub_s),
          Spacer(1, 2 * cm),
          Paragraph(f"Dataset: Amazon Sales Dataset (Kaggle) | {n_products:,} products | "
                    f"{rdf['main_category'].nunique()} main categories", sub_s),
          PageBreak()]

# ----- KPI summary -----
story.append(Paragraph("1. Key Performance Indicators", h1))
kpis = [["Total Products", "Total Revenue (estimated)", "Average Rating"],
        [f"{n_products:,}", f"Rs {total_rev/1e9:.2f} bn", f"{avg_rating:.2f}"],
        ["Average Discount", "Highest Selling Product", "Total Reviews"],
        [f"{avg_disc:.1f}%", best_name, f"{total_reviews/1e6:.1f} M"]]
kt = Table(kpis, colWidths=[5.5 * cm] * 3, rowHeights=[0.8 * cm, 1.2 * cm, 0.8 * cm, 1.2 * cm])
kt.setStyle(TableStyle([
    ("BACKGROUND", (0, 0), (-1, 0), NAVY_C), ("BACKGROUND", (0, 2), (-1, 2), NAVY_C),
    ("TEXTCOLOR", (0, 0), (-1, 0), colors.white), ("TEXTCOLOR", (0, 2), (-1, 2), colors.white),
    ("FONTSIZE", (0, 0), (-1, -1), 10), ("FONTSIZE", (0, 1), (-1, 1), 15), ("FONTSIZE", (0, 3), (-1, 3), 15),
    ("FONTNAME", (0, 1), (-1, 1), "Helvetica-Bold"), ("FONTNAME", (0, 3), (-1, 3), "Helvetica-Bold"),
    ("ALIGN", (0, 0), (-1, -1), "CENTER"), ("VALIGN", (0, 0), (-1, -1), "MIDDLE"),
    ("BOX", (0, 0), (-1, -1), 0.5, colors.HexColor("#D9DEE5")),
    ("INNERGRID", (0, 0), (-1, -1), 0.5, colors.HexColor("#D9DEE5")),
    ("LINEBELOW", (0, 1), (-1, 1), 2, ORANGE_C), ("LINEBELOW", (0, 3), (-1, 3), 2, ORANGE_C),
]))
story += [kt, Spacer(1, 0.6 * cm),
          Paragraph("Note: revenue is estimated as discounted price x rating count, because the dataset "
                    "has no sales quantity column.", cap_s)]

# ----- Category summary table -----
story.append(Paragraph("2. Category Summary", h1))
cat_tbl = (rdf.groupby("main_category")
              .agg(products=("product_id", "count"), revenue=("total_sales", "sum"),
                   rating=("rating", "mean"), discount=("discount_percentage", "mean"))
              .sort_values("revenue", ascending=False))
rows = [["Main Category", "Products", "Revenue (Rs bn)", "Avg Rating", "Avg Discount"]]
for name, r in cat_tbl.iterrows():
    rows.append([name, f"{int(r.products):,}", f"{r.revenue/1e9:.2f}", f"{r.rating:.2f}", f"{r.discount:.1f}%"])
rows.append(["Total", f"{n_products:,}", f"{total_rev/1e9:.2f}", f"{avg_rating:.2f}", f"{avg_disc:.1f}%"])
ct = make_table(rows, [5.2 * cm, 2.4 * cm, 3.4 * cm, 2.6 * cm, 2.8 * cm])
ct.setStyle(TableStyle([("FONTNAME", (0, -1), (-1, -1), "Helvetica-Bold")]))
story += [ct, PageBreak()]

# ----- Statistical summary -----
story.append(Paragraph("3. Statistical Summary", h1))
rows = [["Column", "Mean", "Std", "Min", "Median", "Max"]]
for col in num_cols:
    s = rdf[col]
    rows.append([col, f"{s.mean():,.2f}", f"{s.std():,.2f}", f"{s.min():,.2f}",
                 f"{s.median():,.2f}", f"{s.max():,.2f}"])
story += [make_table(rows, [3.8 * cm, 2.6 * cm, 2.6 * cm, 2.3 * cm, 2.6 * cm, 3.2 * cm], font=7.5),
          Spacer(1, 0.4 * cm),
          Paragraph(f"Data quality: {rdf.shape[0]:,} rows, {rdf.shape[1]} columns, "
                    f"{int(rdf.isna().sum().sum())} missing values, "
                    f"{int(rdf.duplicated().sum())} duplicate rows after cleaning.", body),
          PageBreak()]

# ----- Charts (2 per page, each with a caption) -----
charts = [
    ("01_category_count.png", f"Electronics, Home&Kitchen and Computers&Accessories hold {top3_share:.1f}% of all products."),
    ("02_rating_distribution.png", f"Most products are rated between 3.5 and 4.5; {share_high_rated:.1f}% are rated 4.0 or above."),
    ("03_discount_histogram.png", f"The average discount is {avg_disc:.1f}%, with many products discounted between 30% and 70%."),
    ("04_price_histogram.png", f"The median price is Rs {median_price:,.0f}; {share_under_2000:.1f}% of products cost Rs 2,000 or less."),
    ("05_top_products.png", "The most-reviewed products are low-priced accessories such as HDMI cables and earphones."),
    ("06_rating_vs_price.png", f"Price and rating have a weak relationship (correlation {price_rating_corr:.2f})."),
    ("07_discount_vs_sales.png", f"Higher discounts do not give higher sales (correlation {disc_sales_corr:.2f})."),
    ("08_price_boxplot.png", "Prices are highly skewed: a few expensive products stretch the upper range."),
    ("09_correlation_heatmap.png", "Price, actual price and discount amount move together; rating is almost independent of them."),
    ("10_pair_plot.png", "Pair plot of price, discount, rating and rating count (price and rating count on log scale)."),
]
story.append(Paragraph("4. Charts", h1))
for i, (fname, caption) in enumerate(charts, start=1):
    block = [chart_image(f"report_output/charts/{fname}", max_h=9.2 * cm if fname != "10_pair_plot.png" else 14 * cm),
             Paragraph(f"<b>Figure {i}.</b> {escape(caption)}", cap_s)]
    story.append(KeepTogether(block))
    if i % 2 == 0 and i < len(charts):
        story.append(PageBreak())
story.append(PageBreak())

# ----- Insights and recommendations -----
story.append(Paragraph("5. Business Insights", h1))
for i, line in enumerate(insights, start=1):
    story.append(Paragraph(f"<b>{i}.</b> {escape(line)}", body))
story += [Spacer(1, 0.5 * cm), Paragraph("6. Final Recommendations", h1)]
for i, line in enumerate(recommendations, start=1):
    story.append(Paragraph(f"<b>{i}.</b> {escape(line)}", body))

# ----- Save the PDF -----
pdf_path = "report_output/Amazon_Sales_Report.pdf"
doc = SimpleDocTemplate(pdf_path, pagesize=A4, leftMargin=2 * cm, rightMargin=2 * cm,
                        topMargin=2 * cm, bottomMargin=1.8 * cm,
                        title="Amazon Product Sales & Customer Reviews Analysis")
doc.build(story, onFirstPage=header_footer, onLaterPages=header_footer)
print("PDF saved:", pdf_path)
```
